```python
# Project Develop a Student Performance Tracker using Python, which allows a teacher to input student scores,
# track performance across subjects, and calculate statistics such as averages and feedback on passing or failing,
# using concepts in Object-Oriented Programming (OOP), loops, conditionals, and data structures (lists and dictionaries)

class Student:
    def __init__(self, name):
        self.name = name
        self.scores = {}

    def add_score(self, subject, score):
        if 0 <= score <= 100:
            self.scores[subject] = score
        else:
            print("Invalid score. Score must be between 0 and 100.")

    def get_average_score(self):
        if not self.scores:
            return 0  # Handle case with no scores
        return sum(self.scores.values()) / len(self.scores)

    def get_subject_score(self, subject):
      return self.scores.get(subject, "No score recorded for this subject.")

    def get_feedback(self):
        average_score = self.get_average_score()
        if average_score >= 70:
            return "Passing"
        else:
            return "Failing"


def main():
    students = []
    while True:
        print("\nStudent Performance Tracker Menu:")
        print("1. Add Student")
        print("2. Add Score")
        print("3. View Student Report")
        print("4. Exit")

        choice = input("Enter your choice: ")

        if choice == '1':
            name = input("Enter student name: ")
            students.append(Student(name))
            print(f"Student {name} added.")

        elif choice == '2':
            if not students:
                print("No students added yet.")
                continue

            name = input("Enter student name: ")
            student = next((s for s in students if s.name == name), None)

            if student:
                subject = input("Enter subject: ")
                score = int(input("Enter score: "))
                student.add_score(subject, score)
                print("Score added.")
            else:
                print("Student not found.")

        elif choice == '3':
          if not students:
              print("No students added yet.")
              continue

          name = input("Enter student name: ")
          student = next((s for s in students if s.name == name), None)
          if student:
              print(f"\n--- Report for {student.name} ---")
              if student.scores:
                  for subject, score in student.scores.items():
                      print(f"{subject}: {score}")

                  average = student.get_average_score()
                  print(f"\nAverage Score: {average:.2f}")
                  print(f"Overall Feedback: {student.get_feedback()}")
              else:
                  print("No scores recorded for this student.")
          else:
              print("Student not found.")

        elif choice == '4':
            print("Exiting program.")
            break
        else:
            print("Invalid choice. Please try again.")


if __name__ == "__main__":
    main()
```
